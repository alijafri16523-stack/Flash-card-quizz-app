# New Workspace Setup Guide
This notebook will guide you through setting up a new workspace for a Python project, including environment setup, version control, and initial files.

## 1. Create a New Workspace Directory
Use Python's `os` and `pathlib` modules to create a new directory for your workspace.

```python
import os
from pathlib import Path

# Define the workspace directory name
workspace_dir = Path('my_python_workspace')

# Create the directory if it doesn't exist
workspace_dir.mkdir(exist_ok=True)
print(f"Workspace directory created at: {workspace_dir.resolve()}")
```

## 2. Initialize a Git Repository
Run shell commands from Python to initialize a Git repository in the workspace directory.

```python
import subprocess

# Initialize a git repository in the workspace directory
subprocess.run(['git', 'init'], cwd=workspace_dir)
print(f"Initialized empty Git repository in {workspace_dir.resolve()}/.git/")
```

## 3. Set Up a Python Virtual Environment
Use the `venv` module to create a virtual environment inside the workspace.

```python
import venv

# Path to the virtual environment directory
venv_dir = workspace_dir / 'venv'

# Create the virtual environment
venv.create(venv_dir, with_pip=True)
print(f"Virtual environment created at: {venv_dir.resolve()}")
```

## 4. Install Essential Packages
Use pip to install essential packages such as numpy, pandas, and pytest in the virtual environment.

```python
# Path to pip executable in the virtual environment
pip_executable = venv_dir / 'Scripts' / 'pip.exe' if os.name == 'nt' else venv_dir / 'bin' / 'pip'

# List of essential packages
packages = ['numpy', 'pandas', 'pytest']

# Install packages
subprocess.run([str(pip_executable), 'install'] + packages)
print(f"Installed packages: {', '.join(packages)}")
```

## 5. Configure VS Code Workspace Settings
Create a `.vscode/settings.json` file with recommended Python interpreter and linting settings.

```python
import json

vscode_dir = workspace_dir / '.vscode'
vscode_dir.mkdir(exist_ok=True)
settings_path = vscode_dir / 'settings.json'

settings = {
    "python.pythonPath": str(venv_dir / 'Scripts' / 'python.exe' if os.name == 'nt' else venv_dir / 'bin' / 'python'),
    "python.linting.enabled": True,
    "python.linting.pylintEnabled": True
}

with open(settings_path, 'w') as f:
    json.dump(settings, f, indent=4)
print(f"VS Code settings created at: {settings_path.resolve()}")
```

## 6. Create Initial Project Files
Generate starter files such as `README.md`, `.gitignore`, and a main Python script.

```python
# Create README.md
readme_path = workspace_dir / 'README.md'
with open(readme_path, 'w') as f:
    f.write("""# My Python Workspace\n\nThis is a starter Python project workspace.\n""")

# Create .gitignore
gitignore_path = workspace_dir / '.gitignore'
with open(gitignore_path, 'w') as f:
    f.write("""venv/\n__pycache__/\n*.pyc\n""")

# Create main.py
main_py_path = workspace_dir / 'main.py'
with open(main_py_path, 'w') as f:
    f.write("""def main():\n    print('Hello, world!')\n\nif __name__ == '__main__':\n    main()\n""")

print("Initial project files created: README.md, .gitignore, main.py")
```
